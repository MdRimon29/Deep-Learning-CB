```python
import numpy as np

y_predicted=np.array([1,1,0,0,1])
y_true=np.array([0.30,0.7,1,0,0.5])
```

Using for loop manually

```python
def mse(y_true,y_predicted):
    total_error=0
    for i,j in zip(y_true,y_predicted):
        total_error += (i-j)**2

    print("Total error: ",total_error)
    ms_error=total_error/len(y_predicted)
    print("Mean squared error: ",ms_error)
```

```python
mse(y_true,y_predicted)
```

Using numpy

```python
np.mean((y_true-y_predicted)**2)
```
