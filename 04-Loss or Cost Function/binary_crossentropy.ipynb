```python
def binary_crossentropy(y_true,y_predicted):
    total_error=0
    epsilon=1e-15
    y_predicted_new=np.array([max(i,epsilon) for i in y_predicted])
    y_predicted_new=np.array([min(i,1-epsilon) for i in y_predicted_new])
    for i,j in zip (y_true,y_predicted_new):
        total_error += (i*np.log(j)+(1-i)*np.log(1-j))

    log_loss=-((total_error)/len(y_predicted_new))
    return log_loss
```

Using numpy 

```python
def Binary_crossentropy(y_true,y_predicted):
    epsilon=1e-15
    y_predicted_new=np.array([max(i,epsilon) for i in y_predicted])
    y_predicted_new=np.array([min(i,1-epsilon) for i in y_predicted_new])
    log_loss=-np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))

    return log_loss
```

```python
Binary_crossentropy(y_true,y_predicted)
```
