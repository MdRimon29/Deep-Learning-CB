```python
import numpy as np
import pandas as pd
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df=pd.read_csv("insurance_data.csv")
df.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(df[['age','affordibility']],df.bought_insurance,test_size=0.2,random_state=25)
```

```python
len(X_train)
```

```python
X_train_scaled=X_train.copy()
X_train_scaled['age']=X_train_scaled['age']/100

X_test_scaled=X_test.copy()
X_test_scaled['age']=X_test_scaled['age']/100
```

```python
X_train_scaled.head()
```

```python
model=keras.Sequential([
    keras.layers.Dense(1,input_shape=(2,),activation='sigmoid',kernel_initializer='ones',bias_initializer='zeros')
])
```

```python
model.compile(optimizer='adam',
              loss='binary_crossentropy',
              metrics=['accuracy'])
```

```python
model.fit(X_train_scaled,y_train,epochs=5000)
```

```python
model.evaluate(X_test_scaled,y_test)
```

```python
model.predict(X_test_scaled)
```

```python
coef,intercept=model.get_weights()
coef,intercept
```

```python
def sigmoid(x):
    import math
    return (1/(1+math.exp(-x)))
```

```python
def prediction_function(age,affordibility):
    y=coef[0]*age+coef[1]*affordibility+intercept
    return sigmoid(y)
```

```python
def Binary_crossentropy(y_true,y_predicted):
    epsilon=1e-15
    y_predicted_new=np.array([max(i,epsilon) for i in y_predicted])
    y_predicted_new=np.array([min(i,1-epsilon) for i in y_predicted_new])
    log_loss=-np.mean(y_true*np.log(y_predicted_new)+(1-y_true)*np.log(1-y_predicted_new))

    return log_loss
```

```python
def sigmoid_numpy(x):
    return (1/np.exp(-x))
```

```python
def gradient_descent(age,affordibility,y_true,epochs,min_thresold):
    #output will be -> w1,w2,b
    w1=w2=1
    bias=0
    learning_rate=0.01
    n=len(age)

    for i in range(epochs):
        weighted_sum=w1*age+w2*affordibility+bias
        y_predicted=sigmoid_numpy(weighted_sum)

        loss=Binary_crossentropy(y_true,y_predicted)

        w1d=(1/n)*np.dot((y_predicted-y_true),np.transpose(age))
        w2d=(1/n)*np.dot((y_predicted-y_true),np.transpose(affordibility))

        bias_d=np.mean(y_predicted-y_true)

        w1=w1-learning_rate*w1d
        w2=w2-learning_rate*w2d
        bias=bias-learning_rate*bias_d

        print(f"Epoch:{i}, W1:{w1}, w2:{w2}, b:{bias}, loss:{loss}")

        if(loss<=min_thresold):
            break

    return w1,w2,bias
```
