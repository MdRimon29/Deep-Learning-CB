```python
X=df.drop(60,axis=1)
y=df[60]
```

```python
y=pd.get_dummies(y,drop_first=True,dtype=int)   # here drop_first=True means drops the first category to avoid multicollinearity in regression models or other algorithms
                                                # In our case it drop column for M
y.sample(5)
```

Using Deep Learning

At first create a model without dropout layer

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X,y,train_size=0.8)
```

```python
import tensorflow as tf
from tensorflow import keras

model=keras.Sequential([
    keras.layers.Dense(60,activation='relu'),
    keras.layers.Dense(50,activation='relu'),
    keras.layers.Dense(30,activation='relu'),
    keras.layers.Dense(15,activation='relu'),
    keras.layers.Dense(1,activation='sigmoid'),
])

model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])

model.fit(X_train,y_train,epochs=85)
```

```python
model.evaluate(X_test,y_test)
```

From upper portion we can clearly see that training accuracy>>> test accuracy.

```python
yp=model.predict(X_test)
yp[:10]
```

```python
y_pred=[]

for element in yp:
    if element>=0.5:
        y_pred.append(1)
    else:
        y_pred.append(0)
```
